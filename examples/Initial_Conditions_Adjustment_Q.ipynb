### **Helper Functions**

````python
def gumbel_1(std: float, xbar: float, rp: int or float) -> float:
  """
  Solves the Gumbel Type I probability distribution function (pdf) = exp(-exp(-b)) where b is the covariate. Provide
  the standard deviation and mean of the list of annual maximum flows. Compare scipy.stats.gumbel_r
  Args:
  	std (float): the standard deviation of the series
    xbar (float): the mean of the series
    rp (int or float): the return period in years
  Returns:
  	float, the flow corresponding to the return period specified
  """
  # xbar = statistics.mean(year_max_flow_list)
  # std = statistics.stdev(year_max_flow_list, xbar=xbar)
  return -math.log(-math.log(1 - (1 / rp))) * std * .7797 + xbar - (.45 * std)
````

````python
def fetch_ideam_daily(id_estacion: str, variable: str = 'NV_MEDIA_D', only_approved: bool = False) -> pd.DataFrame:
    """
    Descarga una serie diaria de IDEAM desde S3.

    El bucket publica varias snapshots (parquet) del mismo dataset. Esta
    función las lee todas, deduplica por fecha, y devuelve la versión más
    completa.

    Parámetros
    ----------
    id_estacion : str
        Código de estación con zero-padding (ej. '0026177030').
    variable : str
        Etiqueta IDEAM. Algunas útiles:
          - Nivel:  'NV_MEDIA_D' (medio), 'NV_MN_D' (mínimo), 'NV_MX_D' (máximo)
          - Caudal: 'Q_MEDIA_D'  (medio), 'Q_MN_D'  (mínimo), 'Q_MX_D'  (máximo)
    only_approved : bool
        Si True, filtra `nivel_aprobacion != 'Preliminar'`. Cuidado: deja
        fuera los datos más recientes mientras esperan validación.

    Retorna
    -------
    DataFrame con índice `Fecha` (DatetimeIndex, diario) y una columna que
    se llama 'Nivel (cm)' o 'Caudal (m3/s)' según la variable. Formato
    sparse: los días sin observación no aparecen.
    """
    s3 = boto3.client('s3', endpoint_url='https://datos.ideam.gov.co/', config=Config(signature_version=UNSIGNED), verify=False,)
    prefix = f'observaciones/historicos/parquet/label={variable}/id_estacion={id_estacion}/'
    resp = s3.list_objects_v2(Bucket='s3-estacionesideam', Prefix=prefix)
    files = resp.get('Contents', [])
    if not files:
        raise FileNotFoundError(f"No hay datos en S3 para estación {id_estacion} variable {variable}")
    dfs = []
    for f in files:
        obj = s3.get_object(Bucket='s3-estacionesideam', Key=f['Key'])
        dfs.append(pd.read_parquet(io.BytesIO(obj['Body'].read())))
    df = (pd.concat(dfs, ignore_index=True).drop_duplicates(subset=['fecha'], keep='first').sort_values('fecha').reset_index(drop=True))
    if only_approved:
        df = df[df['nivel_aprobacion'] != 'Preliminar'].copy()

    col_valor = 'Streamflow (m3/s)' if variable.startswith('Q') else 'Water Level (cm)'
    return (df[['fecha', 'valor']].rename(columns={'fecha': 'Datetime', 'valor': col_valor}).set_index('Datetime'))
````

````python
def _parse_section(section: dict, value_key: str) -> pd.DataFrame:
    """
    Parsea una sección del JSON de IDEAM (Hobs, Hsen) y retorna
    un DataFrame con columnas Datetime y Water Level (cm).

    - Filtra solo el sentinel {"ultimo": true}
    - Conserva los valores null como NaN (espacios en blanco)
    - Convierte metros → centímetros
    """
    # value_key in item descarta el sentinel {"ultimo": true}
    records = [item for item in section["data"] if value_key in item]

    df = pd.DataFrame(records)
    df["Fecha"] = pd.to_datetime(df["Fecha"], format="%Y/%m/%d %H:%M")

    df = df.rename(columns={"Fecha": "Datetime", value_key: "Discharge (m3/s)"})
    df.sort_values("Datetime").reset_index(drop=True)

    df.set_index('Datetime', inplace=True)
    df.index = pd.to_datetime(df.index)
    df.index = df.index.to_series().dt.strftime("%Y-%m-%d %H:%M:%S")
    df.index = pd.to_datetime(df.index)

    return df


def fetch_ideam_fews_data(url):
    """
    Retorna (df_obs, df_sen) con todas las filas, conservando NaN.
    Ambas series están en centímetros.
    """
    response = requests.get(url, timeout=30, verify=False)
    response.raise_for_status()
    payload = response.json()

    df_obs = _parse_section(payload["Qobs"], "Qobs")
    df_sen = _parse_section(payload["Qsen"], "Qsen")

    return df_obs, df_sen
````

- **Returns**:

Example of usage:

#### **Step 3: Bias Corrected Forecast**

The `geoglows.bias.correct_forecast` function can be used to perform the MFDC-QM bias correction method on GEOGloWS streamflow forecasts. This is done by transforming each ensemble member using historical flow and observed flow duration curves for the corresponding forecast month.

- **Parameters**:
  - `forecast_data` (pandas.core.frame.DataFrame): A DataFrame containing the forecasted streamflow data (can be forecast records, statistics, or ensemble members) with a datetime index.
  - `simulated_data` (pandas.core.frame.DataFrame): Historical simulated discharge data used to generate the flow duration curve for the month of the forecast.
  - `observed_data` (pandas.core.frame.DataFrame): Discharge observations used to define the monthly stage duration curve.
  - `use_month` (int, optional): Specifies which month's data to use for bias correction:
    - 0: Correct based on the first month of the forecast data (default).
    - 1: Correct based on the last month of the forecast data.
- **Returns**:
  - `pd.DataFrame`:  A bias corrected of the input forecast data, where each streamflow value has been mapped to a corresponding observed discharge value using MFDC-QM.

Example of usage:
```python
corrected_data = geoglows.bias.correct_forecast(df=forecast_ens_df, simulated_data=simulated_data, observed_data=observed_data)
```

This approach assumes that the relationship between discharge and stage remains consistent over time, and leverages monthly duration curves to map forecasted discharge values into corrected values. The method is applied independently to each ensemble member and preserves the probabilistic structure of the forecast.

````python
'''Correct Bias Forecasts'''
def fix_forecast(sim_hist, fore_nofix, obs):
  # print('Calc forecast fixed.')

  # Selection of monthly simulated data
  monthly_simulated = sim_hist[sim_hist.index.month == (fore_nofix.index[0]).month].dropna()

  # Obtain Min and max value
  min_simulated = monthly_simulated.min().values[0]
  max_simulated = monthly_simulated.max().values[0]

  min_factor_df   = fore_nofix.copy()
  max_factor_df   = fore_nofix.copy()
  forecast_ens_df = fore_nofix.copy()

  for column in fore_nofix.columns:
    # Min Factor
    tmp_array = np.ones(fore_nofix[column].shape[0])
    tmp_array[fore_nofix[column] < min_simulated] = 0
    min_factor = np.where(tmp_array == 0, fore_nofix[column] / min_simulated, tmp_array)

    # Max factor
    tmp_array = np.ones(fore_nofix[column].shape[0])
    tmp_array[fore_nofix[column] > max_simulated] = 0
    max_factor = np.where(tmp_array == 0, fore_nofix[column] / max_simulated, tmp_array)

    # Replace
    tmp_fore_nofix = fore_nofix[column].copy()
    tmp_fore_nofix.mask(tmp_fore_nofix <= min_simulated, min_simulated, inplace=True)
    tmp_fore_nofix.mask(tmp_fore_nofix >= max_simulated, max_simulated, inplace=True)

    # Save data
    forecast_ens_df.update(pd.DataFrame(tmp_fore_nofix, index=fore_nofix.index, columns=[column]))
    min_factor_df.update(pd.DataFrame(min_factor, index=fore_nofix.index, columns=[column]))
    max_factor_df.update(pd.DataFrame(max_factor, index=fore_nofix.index, columns=[column]))

  # Get  Bias Correction
  corrected_ensembles = geoglows.bias.correct_forecast(forecast_ens_df, sim_hist, obs)
  corrected_ensembles = corrected_ensembles.multiply(min_factor_df, axis=0)
  corrected_ensembles = corrected_ensembles.multiply(max_factor_df, axis=0)

  return corrected_ensembles
````

#### **Step 4: Bias Corrected Forecast Stats**

This section demonstrates how to compute key statistical measures (like maximum, percentiles, and mean) for the water level transormed forecast ensemble data. The code below uses a Pandas DataFrame to calculate these statistics and organizes them into a structured format for analysis and visualization.

- **Copying and Cleaning Data**:
  - The initial ensemble DataFrame (`corrected_ensembles`) is copied to avoid modifying the original data.
  - `high_res_df` stores a separate high-resolution forecast data column, typically labeled as `ensemble_52`.
  - The `high-resolution` data column is removed from `ensemble`, which will now only contain the remaining ensemble data.
  - Both `ensemble` and `high_res_df` are cleaned to remove any rows with missing values.
- **Calculating Quantiles**:
  - Various quantiles (e.g., max, 75th percentile, 50th percentile, etc.) are calculated from the `ensemble` DataFrame.
  - Each quantile is stored in a separate DataFrame with an appropriate name:
    - `max_df` for the maximum (100th percentile),
    - `p75_df` for the 75th percentile,
    - `p50_df` for the median (50th percentile),
    - `p25_df` for the 25th percentile,
    - `min_df` for the minimum (0th percentile).
- **Calculating Mean**:
  - The mean of all ensemble members is calculated for each time step and renamed to `flow_avg`.
- **Renaming High-Resolution Data**:
  - The high-resolution forecast data is renamed from `ensemble_52` to `high_res`.

### **Forecast Initialization Correction**

````python
start_event = df_obs.index.min().strftime('%Y-%m-%d')
end_event   = df_obs.index.max().strftime('%Y-%m-%d')

daily_dates = pd.date_range(start=start_event, end=end_event, freq='D')

forecasts_dict = {}

for day_date in daily_dates:

    init_str = day_date.strftime("%Y%m%d")
    print(f"Downloading forecast for {init_str}")

    ensemble_df = geoglows.data.forecast_ensembles(river_id=river_id, date=init_str)
    ensemble_df.index = pd.to_datetime(ensemble_df.index)
    ensemble_df[ensemble_df < 0] = 0
    ensemble_df.index = ensemble_df.index.to_series().dt.strftime("%Y-%m-%d %H:%M:%S")
    ensemble_df.index = pd.to_datetime(ensemble_df.index)

    corrected_ensemble_df = fix_forecast(sim_hist=daily_df, fore_nofix=ensemble_df, obs=observed_discharge)

    # ---------------------------------
    # 1️⃣ Separar miembros 3h y 1h
    # ---------------------------------

    # Detect rows where at least one 3h member is non-NaN
    valid_3h_mask = corrected_ensemble_df.iloc[:, :51].notna().any(axis=1)

    df_3h = corrected_ensemble_df.loc[valid_3h_mask, corrected_ensemble_df.columns[:51]]
    df_1h = corrected_ensemble_df.iloc[:, 51:]   # miembro horario

    # ---------------------------------
    # 2️⃣ Convertir a numpy float32
    # ---------------------------------

    ens_3h_array = df_3h.to_numpy(dtype=np.float32)
    ens_1h_array = df_1h.to_numpy(dtype=np.float32)

    # ---------------------------------
    # 3️⃣ Guardar en diccionario
    # ---------------------------------

    forecasts_dict[init_str] = {
        "time_3h": df_3h.index,
        "ens_3h": ens_3h_array,
        "time_1h": df_1h.index,
        "ens_1h": ens_1h_array
    }
````

````python
# === Corrección AR(1) del pronóstico operacional ===
from statsmodels.tsa.arima.model import ARIMA

# 1. Serie de error M52 - H1 en los timestamps de Hobs
init_dates = sorted(forecasts_dict.keys())
init_times = pd.to_datetime(init_dates, format="%Y%m%d")

records = []
for t_obs, h_obs in df_obs.dropna().iloc[:, 0].items():
    valid = init_times <= t_obs
    if not valid.any():
        continue
    last_init = init_times[valid].max()
    fcst = forecasts_dict[last_init.strftime("%Y%m%d")]
    m52 = pd.Series(fcst["ens_1h"].ravel(), index=fcst["time_1h"])
    if t_obs not in m52.index or pd.isna(m52.loc[t_obs]):
        continue
    records.append((t_obs, float(m52.loc[t_obs] - h_obs)))

err = pd.Series(dict(records)).sort_index()

# 2. Ajuste AR(1) — params: [const(=mu), ar.L1(=phi), sigma2]
res = ARIMA(err.values, order=(1, 0, 0)).fit()
mu  = float(res.params[0])
phi = float(res.params[1])

# 3. Error actual y propagación con decay continuo (paso de 12h)
last_obs_time  = df_obs.dropna().index[-1]
last_obs_value = float(df_obs.dropna().iloc[-1, 0])
init_for_enow  = init_times[init_times <= last_obs_time].max()
fcst           = forecasts_dict[init_for_enow.strftime("%Y%m%d")]
m52_at_obs     = pd.Series(fcst["ens_1h"].ravel(),
                           index=fcst["time_1h"]).loc[last_obs_time]
e_now          = float(m52_at_obs - last_obs_value)

elapsed_h = np.array([(t - last_obs_time).total_seconds() / 3600
                      for t in corrected_ensembles.index])
e_hat = mu + (phi ** (elapsed_h / 12.0)) * (e_now - mu)

# 4. Restar la misma corrección a los 52 miembros
e_hat_df = pd.DataFrame(
    np.tile(e_hat.reshape(-1, 1), (1, corrected_ensembles.shape[1])),
    index=corrected_ensembles.index,
    columns=corrected_ensembles.columns
)
corrected_ensembles_ar1 = corrected_ensembles - e_hat_df

display(HTML(corrected_ensembles_ar1.to_html()))
````
